```python
from matplotlib import pyplot as plt
import numpy as np

def f(x):
    return np.sin(x)

x=np.linspace(-5, 10, 1000) #linspace(phi, longueur affichée de l'axe "x", exactitude)

plt.plot(x, f(x), color='red', label="f(x)")
plt.legend()
plt.show
```

```python
x=symbols('x')
def g(x):
    return (x**2+1)/x**3
simplify(diff(g(x)))
```
